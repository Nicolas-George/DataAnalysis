```python
import requests
import pandas as pd
import time

# -------------------------------
# 🔹 1. 기본 설정
# -------------------------------
API_KEY = "75675a79636b6f693531756a4f7066"  # ← 본인의 API 인증키 입력
BASE_URL = "http://openapi.seoul.go.kr:8088"
SERVICE_NAME = "energyUseDataSummaryInfo"
DATA_TYPE = "json"
START_INDEX = 1
END_INDEX = 1000  # 한 번에 최대 1000건
YEARS = range(2015, 2025)
MONTHS = range(1, 13)

# -------------------------------
# 🔹 2. API 호출 함수
# -------------------------------
def get_energy_data(year: int, month: int):
    """특정 연도(year), 월(month)의 에코마일리지 에너지사용량 데이터 요청"""
    url = f"{BASE_URL}/{API_KEY}/{DATA_TYPE}/{SERVICE_NAME}/{START_INDEX}/{END_INDEX}/{year}/{month}"

    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        data = response.json()

        # 데이터 존재 여부 확인
        if SERVICE_NAME in data and "row" in data[SERVICE_NAME]:
            return data[SERVICE_NAME]["row"]
        else:
            print(f"⚠️ 데이터 없음: {year}-{month:02d}")
            return []
    except Exception as e:
        print(f"❌ 요청 실패 ({year}-{month:02d}): {e}")
        return []

# -------------------------------
# 🔹 3. 전체 데이터 수집
# -------------------------------
all_data = []

for year in YEARS:
    for month in MONTHS:
        records = get_energy_data(year, month)
        if records:
            all_data.extend(records)
        time.sleep(0.3)  # API 호출 간 간격 (서버 부하 방지)

# -------------------------------
# 🔹 4. DataFrame 변환
# -------------------------------
df = pd.DataFrame(all_data)
print(f"총 데이터 개수: {len(df)}")

# -------------------------------
# 🔹 5. 개인 유형만 필터링
# -------------------------------
# ⚠️ 실제 API 데이터 컬럼명 확인 필요 (예: GUBUN_NM, MEMBER_TYPE 등)
if 'MM_TYPE' in df.columns:
    df_personal = df[df['MM_TYPE'].str.contains('개인', na=False)].copy()
else:
    print("⚠️ 'MM_TYPE' 컬럼이 없습니다. 실제 데이터 컬럼명을 확인해주세요.")
    df_personal = df.copy()

print(f"개인 유형 데이터 개수: {len(df_personal)}")

# -------------------------------
# 🔹 6. 필요한 컬럼만 추출
# -------------------------------
# ⚠️ 실제 컬럼명 예시: BASE_YM, ELEC_USE, GAS_USE, WATER_USE, HEAT_USE
columns_needed = [
    "YEAR",      # 기준연월
    "MM_TYPE",     # 회원유형
    "EUS",     # 전기 사용량
    "GUS",      # 가스 사용량
    "WUS",    # 수도 사용량
    "HUS"      # 지역난방 사용량
]

existing_cols = [col for col in columns_needed if col in df_personal.columns]
df_energy = df_personal[existing_cols].copy()

# -------------------------------
# 🔹 7. CSV 저장
# -------------------------------
output_file = "eco_energy_personal_2015_2024.csv"
df_energy.to_csv(output_file, index=False, encoding="utf-8-sig")

print(f"✅ 개인 유형 전기·가스·수도·지역난방 데이터 저장 완료: {output_file}")
```

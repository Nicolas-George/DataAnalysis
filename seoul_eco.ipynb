```python
import requests
import pandas as pd
import time

# -------------------------------
# 🔹 1. 기본 설정
# -------------------------------
API_KEY = "75675a79636b6f693531756a4f7066"  # ← 본인의 API 인증키 입력
BASE_URL = "http://openapi.seoul.go.kr:8088"
SERVICE_NAME = "energyUseDataSummaryInfo"
DATA_TYPE = "json"
START_INDEX = 1
END_INDEX = 1000  # 한 번에 최대 1000건
YEARS = range(2015, 2025)
MONTHS = range(1, 13)

# -------------------------------
# 🔹 2. API 호출 함수
# -------------------------------
def get_energy_data(year: int, month: int):
    """특정 연도(year), 월(month)의 에코마일리지 에너지사용량 데이터 요청"""
    url = f"{BASE_URL}/{API_KEY}/{DATA_TYPE}/{SERVICE_NAME}/{START_INDEX}/{END_INDEX}/{year}/{month}"

    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        data = response.json()

        # 데이터 존재 여부 확인
        if SERVICE_NAME in data and "row" in data[SERVICE_NAME]:
            return data[SERVICE_NAME]["row"]
        else:
            print(f"⚠️ 데이터 없음: {year}-{month:02d}")
            return []
    except Exception as e:
        print(f"❌ 요청 실패 ({year}-{month:02d}): {e}")
        return []

# -------------------------------
# 🔹 3. 전체 데이터 수집
# -------------------------------
all_data = []

for year in YEARS:
    for month in MONTHS:
        records = get_energy_data(year, month)
        if records:
            all_data.extend(records)
        time.sleep(0.3)  # API 호출 간 간격 (서버 부하 방지)

# -------------------------------
# 🔹 4. DataFrame 변환
# -------------------------------
df = pd.DataFrame(all_data)
print(f"총 데이터 개수: {len(df)}")

# -------------------------------
# 🔹 5. 개인 유형만 필터링
# -------------------------------
# ⚠️ 실제 API 데이터 컬럼명 확인 필요 (예: GUBUN_NM, MEMBER_TYPE 등)
if 'MM_TYPE' in df.columns:
    df_personal = df[df['MM_TYPE'].str.contains('개인', na=False)].copy()
else:
    print("⚠️ 'MM_TYPE' 컬럼이 없습니다. 실제 데이터 컬럼명을 확인해주세요.")
    df_personal = df.copy()

print(f"개인 유형 데이터 개수: {len(df_personal)}")

# -------------------------------
# 🔹 6. 필요한 컬럼만 추출
# -------------------------------
# ⚠️ 실제 컬럼명 예시: BASE_YM, ELEC_USE, GAS_USE, WATER_USE, HEAT_USE
columns_needed = [
    "YEAR",      # 기준연월
    "MM_TYPE",     # 회원유형
    "EUS",     # 전기 사용량
    "GUS",      # 가스 사용량
    "WUS",    # 수도 사용량
    "HUS"      # 지역난방 사용량
]

existing_cols = [col for col in columns_needed if col in df_personal.columns]
df_energy = df_personal[existing_cols].copy()

# -------------------------------
# 🔹 7. CSV 저장
# -------------------------------
output_file = "eco_energy_personal_2015_2024.csv"
df_energy.to_csv(output_file, index=False, encoding="utf-8-sig")

print(f"✅ 개인 유형 전기·가스·수도·지역난방 데이터 저장 완료: {output_file}")
```

```python
import requests
import pandas as pd
from tqdm import tqdm  # 진행률 표시용

# 🔑 본인의 API 키 입력
API_KEY = "75675a79636b6f693531756a4f7066"

# 기본 URL
BASE_URL = "http://openapi.seoul.go.kr:8088"

# 수집 연도 및 월 범위 설정
years = range(2015, 2025)
months = range(1, 13)

# 결과 저장용 리스트
data_list = []

# 데이터 수집
for year in tqdm(years, desc="연도별 진행"):
    for month in months:
        url = f"{BASE_URL}/{API_KEY}/json/energyUseDataSummaryInfo/1/1000/{year}/{month:02d}"

        try:
            response = requests.get(url)
            response.raise_for_status()
            data = response.json()

            # API 응답 데이터 확인
            rows = data.get("energyUseDataSummaryInfo", {}).get("row", [])
            if not rows:
                continue

            # 개인 유형 + 에너지 종류 필터링
            for row in rows:
                if row.get("M_TYPE") == "개인" and row.get("ENERGY_GB") in ["전기", "가스", "수도", "지역난방"]:
                    data_list.append({
                        "연도": year,
                        "월": month,
                        "회원유형": row.get("M_TYPE"),
                        "에너지종류": row.get("ENERGY_GB"),
                        "사용량": row.get("USE_AMT"),
                        "단위": row.get("UNIT"),
                        "구분": row.get("GUBUN"),
                        "지역": row.get("GU_NM")
                    })

        except Exception as e:
            print(f"[오류] {year}-{month:02d} 데이터 수집 실패: {e}")

# pandas DataFrame으로 변환
df = pd.DataFrame(data_list)

# 결과 저장
df.to_csv("eco_energy_personal_2015_2024.csv", index=False, encoding="utf-8-sig")

print(f"\n✅ 데이터 수집 완료! 총 {len(df)}건 저장됨")
print("📂 파일명: eco_energy_personal_2015_2024.csv")
```

```python
import requests
import json
import time

# === 서울시 OPEN API 기본 설정 ===
API_KEY = "75675a79636b6f693531756a4f7066"  # 👉 본인 API 키 입력
BASE_URL = "http://openapi.seoul.go.kr:8088"

# === 데이터 수집 범위 설정 ===
START_YEAR = 2015
END_YEAR = 2024

# === 수집할 에너지 종류 ===
ENERGY_TYPES = ["전기", "가스", "수도", "지역난방"]

# === 결과 저장용 리스트 ===
all_data = []

def get_energy_data(year, month):
    """
    주어진 연도와 월에 대해 API 호출을 수행하고,
    개인(회원유형: 개인)의 에너지 사용량 데이터를 반환.
    """
    url = f"{BASE_URL}/{API_KEY}/json/energyUseDataSummaryInfo/1/1000/{year}/{month:02d}"
    response = requests.get(url)

    # API 응답 코드 확인
    if response.status_code != 200:
        print(f"[오류] {year}-{month:02d} API 호출 실패 (HTTP {response.status_code})")
        return None

    data = response.json()

    # API 호출 성공 여부 확인
    if "energyUseDataSummaryInfo" not in data:
        print(f"[오류] {year}-{month:02d} 데이터 없음 또는 API 구조 변경됨")
        return None

    rows = data["energyUseDataSummaryInfo"].get("row", [])

    # 개인 유형만 필터링
    personal_data = [r for r in rows if r.get("GUBUN_NM") == "개인"]

    # 에너지별 데이터 추출
    filtered = [
        {
            "년도": year,
            "월": month,
            "회원유형": r.get("GUBUN_NM"),
            "에너지원": r.get("ENERGY_GB"),
            "사용량": r.get("USE_AMT"),
            "단위": r.get("UNIT")
        }
        for r in personal_data if r.get("ENERGY_GB") in ENERGY_TYPES
    ]

    return filtered

# === 전체 데이터 수집 ===
print("서울시 에코마일리지 에너지사용량 데이터 수집을 시작합니다...\n")

for year in range(START_YEAR, END_YEAR + 1):
    for month in range(1, 13):
        data = get_energy_data(year, month)
        if data:
            all_data.extend(data)
            print(f"[성공] {year}-{month:02d} 데이터 {len(data)}건 수집 완료")
        else:
            print(f"[경고] {year}-{month:02d} 데이터 없음")
        time.sleep(0.3)  # 서버 과부하 방지를 위해 잠시 대기

print("\n✅ 데이터 수집 완료!")
print(f"총 {len(all_data)}건의 데이터가 수집되었습니다.")

# === 결과 JSON 파일로 저장 ===
output_filename = "eco_mileage_personal_energy_2015_2024.json"
with open(output_filename, "w", encoding="utf-8") as f:
    json.dump(all_data, f, ensure_ascii=False, indent=2)

print(f"\n💾 결과가 '{output_filename}' 파일로 저장되었습니다.")
```
